```python
"""
prompt: image_dir = "/mnt/abka03/xlvlm_data/cifar_100_samples/val", from this chose randomly 16 sub directoris and chose randomly 16 images from each of the 16 subdirectroies. and create a new image with this 16 image 4 by 4 grid . name this imaage as a a randormnumber_all the16 subdirs nam , spalce removed with underscore. sample N times and store all of them in a new output directory."
"""
import os
import random
from PIL import Image
import uuid

# Define the image directory and output directory
image_dir = "/mnt/abka03/xlvlm_data/imagenet_1000_concepts/val"
output_dir = "/mnt/abka03/xlvlm_data/imagenet_1000_concepts/val_sampled_grids"
os.makedirs(output_dir, exist_ok=True)

# Function to create a 4x4 grid image
def create_image_grid(image_paths, output_path):
    images = [Image.open(img_path) for img_path in image_paths]
    grid_size = 4
    img_width, img_height = images[0].size
    grid_img = Image.new('RGB', (grid_size * img_width, grid_size * img_height))

    for i, img in enumerate(images):
        x = (i % grid_size) * img_width
        y = (i // grid_size) * img_height
        grid_img.paste(img, (x, y))

    grid_img.save(output_path)

# Function to sample N grids
def sample_grids(image_dir, output_dir, N):
    subdirs = [d for d in os.listdir(image_dir) if os.path.isdir(os.path.join(image_dir, d))]
    for _ in range(N):
        selected_subdirs = random.sample(subdirs, 16)
        image_paths = []
        for subdir in selected_subdirs:
            subdir_path = os.path.join(image_dir, subdir)
            images = [os.path.join(subdir_path, img) for img in os.listdir(subdir_path) if img.endswith(('.png', '.jpg', '.jpeg'))]
            image_paths.append(random.choice(images))
        
        # Generate a unique name for the output image
        subdirs_name = "_".join([subdir.replace(" ", "_") for subdir in selected_subdirs])
        output_image_name = f"{uuid.uuid4().hex}_{subdirs_name}.jpg"
        output_path = os.path.join(output_dir, output_image_name)
        
        # Create the 4x4 grid image
        create_image_grid(image_paths, output_path)

# Sample N grids
N = 10  # Change this to the desired number of samples
sample_grids(image_dir, output_dir, N)
```
