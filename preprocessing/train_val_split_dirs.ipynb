```python
import os
import shutil
import random

source_dir = '/mnt/abka03/xlvlm_data/dtd'
train_dir = '/mnt/abka03/xlvlm_data/dtd_processed/train'
val_dir = '/mnt/abka03/xlvlm_data/dtd_processed/val'
split_ratio = 0.8

# Create train and val directories
os.makedirs(train_dir, exist_ok=True)
os.makedirs(val_dir, exist_ok=True)

for class_name in os.listdir(source_dir):
    class_path = os.path.join(source_dir, class_name)
    if os.path.isdir(class_path):
        images = os.listdir(class_path)
        random.shuffle(images)

        split_point = int(len(images) * split_ratio)
        train_images = images[:split_point]
        val_images = images[split_point:]

        # Create class directories
        train_class_dir = os.path.join(train_dir, class_name)
        val_class_dir = os.path.join(val_dir, class_name)
        os.makedirs(train_class_dir, exist_ok=True)
        os.makedirs(val_class_dir, exist_ok=True)

        # Move images
        for img in train_images:
            shutil.copy2(os.path.join(class_path, img), os.path.join(train_class_dir, img))
        for img in val_images:
            shutil.copy2(os.path.join(class_path, img), os.path.join(val_class_dir, img))

print("Dataset split completed.")
```

```python
import os
import shutil
import random

train_dir = '/mnt/abka03/xlvlm_data/dtd_processed/train'
output_dir = '/mnt/abka03/xlvlm_data/dtd_processed/train_noisy'

os.makedirs(output_dir, exist_ok=True)

train_classes = [cls for cls in os.listdir(train_dir) if os.path.isdir(os.path.join(train_dir, cls))]

for class_name in train_classes:
    class_path = os.path.join(train_dir, class_name)
    output_class_path = os.path.join(output_dir, class_name)
    os.makedirs(output_class_path, exist_ok=True)

    # Clean images
    clean_images = os.listdir(class_path)
    random.shuffle(clean_images)
    clean_count = len(clean_images) // 2
    selected_clean = clean_images[:clean_count]

    # Noisy images from other classes
    noisy_images = []
    for other_class in train_classes:
        if other_class == class_name:
            continue
        other_class_path = os.path.join(train_dir, other_class)
        noisy_images.extend([
            os.path.join(other_class_path, img) for img in os.listdir(other_class_path)
        ])

    random.shuffle(noisy_images)
    selected_noisy = noisy_images[:clean_count]

    # Copy clean images
    for img in selected_clean:
        src = os.path.join(class_path, img)
        dst = os.path.join(output_class_path, img)
        shutil.copy2(src, dst)

    # Copy noisy images with prefix
    for idx, img_path in enumerate(selected_noisy):
        dst = os.path.join(output_class_path, f"noisy_{idx}_{os.path.basename(img_path)}")
        shutil.copy2(img_path, dst)

    print(f"✅ {class_name}: {clean_count} clean + {clean_count} noisy from other classes.")

print(f"\n✅ Done. Mixed dataset saved in '{output_dir}'.")
```

```python
import os

def count_jpg_in_subdirs(root_dir):
    min_count = None
    min_subdir = None

    for dirpath, dirnames, filenames in os.walk(root_dir):
        if dirpath == root_dir:
            continue  # Skip the root directory itself

        jpg_count = sum(1 for file in filenames if file.lower().endswith('.jpg'))

        if min_count is None or jpg_count < min_count:
            min_count = jpg_count
            min_subdir = dirpath

    if min_count is not None:
        print(f"Minimum .jpg files: {min_count} in '{min_subdir}'")
    else:
        print("No subdirectories with .jpg files found.")

# Example usage
count_jpg_in_subdirs('/mnt/abka03/xlvlm_data/dtd_processed/val')
```
