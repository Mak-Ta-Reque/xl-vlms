```python
import kagglehub
import shutil
import os

# Step 1: Download the dataset using kagglehub
path = kagglehub.dataset_download("titericz/imagenet1k-val")

print("Downloaded dataset path:", path)

# Step 2: Define your target directory
target_path = "/mnt/abka03/raw_data_download/imagent"

# Step 3: Create target directory if it doesn't exist
os.makedirs(target_path, exist_ok=True)

# Step 4: Move downloaded files to your target directory
for item in os.listdir(path):
    s = os.path.join(path, item)
    d = os.path.join(target_path, item)
    if os.path.isdir(s):
        shutil.copytree(s, d, dirs_exist_ok=True)
    else:
        shutil.copy2(s, d)

print("Files successfully moved to:", target_path)
```

## Rename word-Net names

```python
import os

# Path to your mapping text file
mapping_file = "/mnt/abka03/Projects/xl-vlms/preprocessing/imagenet_names.txt"

# Path to the dataset folder with WordNet ID directories
dataset_dir = "/mnt/abka03/raw_data_download/train"

# Parse the mapping file
wnid_to_classname = {}
with open(mapping_file, 'r') as f:
    for line in f:
        parts = line.strip().split()
        if len(parts) >= 3:
            wnid = parts[0]
            class_name = "_".join(parts[2:])  # Handles multi-word class names
            wnid_to_classname[wnid] = class_name

# Rename directories
for dir_name in os.listdir(dataset_dir):
    old_path = os.path.join(dataset_dir, dir_name)
    if os.path.isdir(old_path) and dir_name in wnid_to_classname:
        new_dir_name = wnid_to_classname[dir_name]
        new_path = os.path.join(dataset_dir, new_dir_name)
        if not os.path.exists(new_path):
            os.rename(old_path, new_path)
            print(f"Renamed {dir_name} → {new_dir_name}")
        else:
            print(f"Skipped {dir_name}: target name '{new_dir_name}' already exists.")
```
